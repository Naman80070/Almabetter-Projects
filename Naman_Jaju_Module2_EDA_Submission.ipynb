Data Cleaning & Feature Engineering

The dataset is largely clean (no missing values or duplicate rows were found), but a few structural improvements make the analysis more meaningful:

1. Several count-based columns (e.g., `Num_Bank_Accounts`, `Num_Credit_Card`) are stored as `float64` — converting them to `int` improves readability without losing information.
2. `Type_of_Loan` is a comma-separated string listing multiple loan types per customer — we engineer a new numeric feature `Num_Loan_Types` capturing how many distinct loan types a customer holds.
3. Identifier columns (`ID`, `Customer_ID`, `Name`, `SSN`) carry no predictive/analytical value and are excluded from the visualization dataframe (kept in the original `df` for traceability).


````python
# Columns that are logically whole numbers (counts) but stored as float
int_like_cols = [
    'Age', 'Num_Bank_Accounts', 'Num_Credit_Card', 'Interest_Rate',
    'Num_of_Loan', 'Delay_from_due_date', 'Num_of_Delayed_Payment',
    'Num_Credit_Inquiries', 'Credit_History_Age'
]

for col in int_like_cols:
    try:
        df[col] = df[col].round(0).astype(int)
    except Exception as e:
        print(f"Could not convert {col}: {e}")

print("Data types after conversion:")
print(df[int_like_cols].dtypes)
````

````python
# FEATURE ENGINEERING: Number of distinct loan types per customer

def count_loan_types(loan_str):
    """Counts the number of distinct loan types listed in the Type_of_Loan string."""
    if pd.isna(loan_str) or loan_str.strip().lower() == 'not specified':
        return 0
    # Loan types are separated by commas and the word 'and'
    cleaned = loan_str.replace(' and ', ', ')
    loans = [l.strip() for l in cleaned.split(',') if l.strip() != '']
    return len(loans)

df['Num_Loan_Types'] = df['Type_of_Loan'].apply(count_loan_types)
df[['Type_of_Loan', 'Num_Loan_Types']].head()
````

````python
# Drop identifier columns that carry no analytical value for EDA/visualization
id_cols = ['ID', 'Customer_ID', 'Name', 'SSN']
df_eda = df.drop(columns=id_cols)

print("Columns available for analysis:")
print(df_eda.columns.tolist())
````

````python
# Order the target variable's categories logically for consistent plotting
credit_score_order = ['Poor', 'Standard', 'Good']
df_eda['Credit_Score'] = pd.Categorical(df_eda['Credit_Score'], categories=credit_score_order, ordered=True)

# Define a consistent color palette for the target variable across all charts
score_palette = {'Poor': '#E74C3C', 'Standard': '#F1C40F', 'Good': '#2ECC71'}
````
